```python

import pandas as pd
from pathlib import Path

# Load dataset
df = pd.read_csv('indian_roads_dataset.csv')

# 1. Basic information
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

# 2. Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Statistical summary
print("\nStatistical Summary:")
print(df.describe(include='all'))

# 4. Total accidents by state
print("\nAccidents by State:")
print(df['state'].value_counts())

# 5. Total accidents by cause
print("\nAccidents by Cause:")
print(df['cause'].value_counts())

# 6. Accident severity analysis
print("\nAccidents by Severity:")
print(df['accident_severity'].value_counts())

# 7. Accidents by weather
print("\nAccidents by Weather:")
print(df['weather'].value_counts())

# 8. Accidents by hour
print("\nAccidents by Hour:")
print(df['hour'].value_counts().sort_index())

# 9. Accidents by road type
print("\nAccidents by Road Type:")
print(df['road_type'].value_counts())

# 10. Monthly accident analysis
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['month'] = df['date'].dt.month

print("\nAccidents by Month:")
print(df.groupby('month').size())

# 11. Total casualties by state
print("\nTotal Casualties by State:")
print(df.groupby('state')['casualties'].sum().sort_values(ascending=False))

# 12. Average risk score by road type
print("\nAverage Risk Score by Road Type:")
print(df.groupby('road_type')['risk_score'].mean())

# 13. Overall results
print("\nTotal Accident Records:", len(df))
print("Total Casualties:", df['casualties'].sum())
print("Average Risk Score:", df['risk_score'].mean())

print("\nData Analysis Completed!")
```
